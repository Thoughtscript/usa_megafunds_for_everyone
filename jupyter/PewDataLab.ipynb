```python
import pandas as pd
import numpy as np
import csv
```

```python
FILE_PATH="_data/Pew-Total-Balances-Data.csv"
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

```python
RAW_DATA = []
ROWS_DATA = []
with open(FILE_PATH, mode='r', newline='', encoding='utf-8') as f:
    reader = csv.reader(f)
    for row in reader:
        RAW_DATA.append(row)
        if (len(row)==163):
            ROWS_DATA.append(row)
            
```

```python
print(ROWS_DATA)
```

```python
df = pd.DataFrame(ROWS_DATA)
df.columns = df.iloc[0]
df = df[1:].reset_index(drop=True)
print(df)
```

```python
list(df)
```

```python
RAINY_DAY_2026_COL = "Rainy day funds: balances 2026"
TOTAL_BAL_2026_COL = "Total balances: balances 2026"
RAINY_DAY_2026_PER_SPENDING_COL = "Rainy day funds: % of state spending 2026"
TOTAL_BAL_2026_PER_SPENDING_COL = "Total balances: % of state spending 2026"
```

```python
def prep_and_show_dollars(COL):
    to_num = pd.to_numeric(df[COL], errors='coerce')
    to_num_mean = to_num.mean()
    print(COL + " mean:")
    print(f"${to_num_mean:,.2f}")

    to_num_median = to_num.median()
    print(COL + " median:")
    print(f"${to_num_median:,.2f}")
    
def prep_and_show_percent(COL):
    to_num = pd.to_numeric(df[COL], errors='coerce')
    
    to_num_mean = to_num.mean() * 100
    print(COL)
    print(f"{to_num_mean:,.2f}%")

    to_num_median = to_num.median() * 100
    print(COL)
    print(f"{to_num_median:,.2f}%")
```

```python
prep_and_show_dollars(RAINY_DAY_2026_COL)
```

```python
prep_and_show_dollars(TOTAL_BAL_2026_COL)
```

```python
prep_and_show_percent(RAINY_DAY_2026_PER_SPENDING_COL)
```

```python
prep_and_show_percent(TOTAL_BAL_2026_PER_SPENDING_COL)
```
